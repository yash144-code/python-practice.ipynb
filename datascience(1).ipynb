```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df= pd.read_csv("Sales Dataset.csv")
sns.pairplot(df,vars = ["Age","Quantity","Total Amount"],hue= "Gender")
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df= pd.read_csv("Sales Dataset.csv")
corr = df.corr(numeric_only = True)
corr = corr.drop("Unnamed: 0",axis = 1)
print(corr)
plt.figure(figsize=(8,6))

sns.heatmap(corr,annot=True,cmap="Blues")
# cmap="coolwarm", cmap="viridis", cmap="Blues" , cmap="Reds"
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df= pd.read_csv("Sales Dataset.csv")
corre = df[["Quantity","Total Amount"]].corr()
print(corre)
sns.heatmap(corre,annot=True,cmap = "coolwarm")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df= pd.read_csv("Sales Dataset.csv")
sns.regplot(data =df,x="Quantity",y= "Total Amount",ci = None)
plt.title("Quantity Vs Total amount")
plt.show()
```
